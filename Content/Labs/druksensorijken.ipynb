Een soortgelijke meettechniek gaan we gebruiken om een druksensor te ijken. Van de sensor zijn wel wat dingen bekend, maar omdat de spanning van de Arduino niet overeenkomstig is met de gewenste spanning, zouden we deze moeten ijken. We weten dat de sensor lineair is, dus als we twee punten weten, kunnen we de rest van de curve bepalen. Nog beter zou het zijn om drie punten te nemen en zo ook het lineaire karakter van de sensor te bevestigen.

## Theorie

Een injectiespuit met een maximaal volume van 50 mL is gevuld met lucht. De spuit kan aan een kant afgesloten worden met een tube die verbonden is met een druksensor die de gasdruk meet. Door de zuiger van de spuit in te drukken, wordt het volume verkleind en de druk verhoogd. Wanneer we de druk langzaam in drukken verwachten we dat de druk in de spuit volgens de wet van Boyle toeneemt:

$$
    P_1 V_1 = P_2 V_2 
$$ (eq:Boyle)

Omdat de gemeten spanning van de druksensor lineair afhankelijk is van de druk, kan de druk uitgedrukt worden als:

$$
    P = a U + b
$$ (eq:lineair)

## Resultaten

````python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

#data van deel 1
U_1 = 179 #bits bij 40 ml
U_2 = 309 #bits bij 20 ml
U_3 = 536 #bits bij 10 ml
bits = np.array([U_1, U_2, U_3])

P_1 = 1 #bar bij 40 ml
P_2 = 2 #bar bij 20 ml
P_3 = 4 #bar bij 10 ml
pressures = np.array([P_1, P_2, P_3])

def bit_to_P(U, a, b):
    return a * U + b

popt, cov = curve_fit(bit_to_P, bits, pressures)

a, b = popt
print(a,b)

plt.figure()
plt.scatter(bits, pressures, label='Data punten')
plt.plot(bits, bit_to_P(bits, a, b))

````

````python
#metingen met lange buis
#in bits
U_lb1 = 179
U_lb2 = 203
U_lb3 = 233
U_lb4 = 279
U_lb5 = 356
U_lb6 = 161
U_lb7 = 147
U_lb8 = 136
U_lb9 = 126
U_lb10 = 119
#volumes in ml
V_1 = 30 
V_2 = 25
V_3 = 20
V_4 = 15
V_5 = 10
V_6 = 35
V_7 = 40
V_8 = 45
V_9 = 50
V_10 = 55

volumes_cil = np.array([V_1, V_2, V_3, V_4, V_5, V_6, V_7, V_8, V_9, V_10]) 
bits_lb = np.array([U_lb1, U_lb2, U_lb3, U_lb4, U_lb5, U_lb6, U_lb7, U_lb8, U_lb9, U_lb10])
druk = a*bits_lb + b

def func_druk(V,k,V_d):
    return k/(V+V_d)

popt_lb, cov_lb = curve_fit(func_druk, volumes_cil, druk)
k, d = popt_lb
k_u, d_u = np.sqrt(np.diag(cov_lb))
print(k,d)
print("Onzekerheid k:", k_u)
print("Onzekerheid dode ruimte:", d_u)
print("De dode ruimte van het lange buisje is:", d, "+/-", d_u, "cm^3")


#andere niet werkende manier om V_d te berekenen

#V_d = popt_lb[1]
#rint("De dode ruimte is:", V_d, "cm^3")
#druk = druk/100000  #omzetten naar bar
#V_tot = volumes_cil + V_d

#def func_tot(k,V_tot):
    #return k/V_tot

#x = np.linspace(min(V_tot),max(V_tot),10)
#y = func_tot(k,V_tot)

#plt.figure
#Plt.scatter(V_tot, druk, label='Lange buis metingen')
#plt.plot(x, y, "r--", label='Fit: k=%.2f' % (k))
##plt.xlabel('Totaal Volume (ml)')
#plt.ylabel('Druk (bar)')
#plt.legend()
#plt.show()

#mijn niet werkende manieren om de grafiek te plotten

x_test = np.linspace(0.8*np.min(volumes_cil),1.2*np.max(volumes_cil),1000)
y_test = func_druk(x_test,*popt_lb)

plt.figure()
plt.scatter(volumes_cil, druk, label='Lange buis metingen')
plt.plot(x_test, y_test, label='Fit: k=%.2f, d=%.2f' % (k,d))
plt.xlabel('Volume (ml)')
plt.ylabel('Druk (Pa)')
plt.legend()
plt.show()
````
